```python
import os
from google.colab import userdata
from langchain_groq import ChatGroq

# Get API key from Colab Secrets
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

# Initialize Groq LLM
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

# System prompt
SYSTEM = """
You are the assistant for Sweet Crumbs Home Bakery.

Facts:
- We bake to order.
- Orders need 24 hours' notice.
- All cakes have eggless versions.

Rules:
- Never promise delivery times.
- If you do not know, say so.
- Reply politely in under 60 words.
"""

# Ask function
def ask(question):
    return llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ]).content

# Test
print(ask("Can I get my order tomorrow?"))
```

```python
SYSTEM = """You are the assistant for Sweet Crumbs Home Bakery.
Facts: we bake to order; orders need 24 hours' notice; all cakes have eggless versions.
Rules: never promise delivery times; if you do not know, say so.
Format: reply politely in under 60 words."""
```

```python
def ask(question):
    return llm.invoke([("system", SYSTEM), ("human", question)]).content
```

```python
# Reinitialize Groq with the updated model
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

# Recreate the ask function
def ask(question):
    response = llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ])
    return response.content

# Verify active model
print("Active model:", llm.model_name)

# Test
print(ask("Can I get my order tomorrow?"))
```
